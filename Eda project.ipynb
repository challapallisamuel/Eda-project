```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(response.text)
```

```python
find -> fetch the first occuraence information of that tag
find_all -> fetch all the the information of that tag
```

```python
soup.find("div",class_="KzDlHZ")
```

```python
soup.find("div",class_="KzDlHZ").text
```

```python
soup.find_all("div",class_="KzDlHZ")
```

```python
for i in soup.find_all("div",class_="KzDlHZ"):
    print(i.text)
```

```python
soup.find("div",class_="KzDlHZ").text
```

```python
import re
```

```python
re.findall("^(\w+)\s",soup.find("div",class_="KzDlHZ").text)
```

```python
brand = []
for i in soup.find_all("div",class_="KzDlHZ"):
    laptop = re.findall("^(\w+)\s",i.text)[0]
    brand.append(laptop)
```
